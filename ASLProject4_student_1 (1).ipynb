##### Create a function called `sigmoid(x)`.

This will be one of your **activation functions** that takes some vector `x` and returns the sigmoid of the vector. The `sigmoid` function can be written as:

\begin{gather}
  \sigma(x) = \frac{1}{1 + e^{-x}}
\end{gather}

```python
def sigmoid(x):
  # Apply the sigmoid function element-wise
    return 1 / (1 + np.exp(-x))
```

##### Create a function called `sigmoid_derivative(x)`.

This function will calculate the derivative of the activation function for backpropogation in the network at some point, `x`. The function should return the derivative of the sigmoid function at that point. The derivative of the Sigmoid function is written as:

\begin{gather}
  \sigma^\prime (x) = \sigma(x) * (1 - \sigma(x))
\end{gather}

```python

def sigmoid_derivative(x):
  # Compute sigmoid first
    s = sigmoid(x)

    # Apply derivative formula: σ(x) * (1 - σ(x))
    return s * (1 - s)
```

##### Create a function called `relu(x)`.

This function will calculate and return the `ReLU` of some input value `x`. The ReLU function can be written as:

\begin{gather}
  \sigma(x) = \begin{cases} x, & x > 0 \\ 0, & x \leq 0 \end{cases} = max(0, x)
\end{gather}

```python

def relu(x):
  # Apply ReLU element-wise: max(0, x)
    return np.maximum(0, x)
```

##### Create a function called `relu_derivative(x)`.

This will calculate the derivative of the `ReLU` function for the given input. The derivative of ReLU can be written as:

\begin{gather}
  \sigma^{\prime}(x) = \begin{cases} 1, & x > 0 \\ 0, & x \leq 0 \end{cases}
\end{gather}

```python

def relu_derivative(x):
  # Derivative is 1 where x > 0, else 0
    return (x > 0).astype(float)
```

##### Create a new function called `tanh()`.

There are other activation functions than just Sigmoid and ReLU. Note: for this question, and this question only, you are restricted to only using the `np.exp()` function from Numpy. Do **not** directly use `np.tanh()` for this question.

Hint: The equation for the hyperbolic tangent is:

\begin{gather}
  \sigma(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
\end{gather}

```python

def tanh(x):
  # Compute tanh using only np.exp()
    exp_pos = np.exp(x)
    exp_neg = np.exp(-x)

    return (exp_pos - exp_neg) / (exp_pos + exp_neg)
```

##### Define a `tanh_derivative()` function.

The derivative of the hyperbolic tangent is the **square of the hyperbolic secant**.  
Do **not** directly use `np.tanh()` for this question.


you can write the derivative as

$$
\frac{d}{dx}\tanh(x)=\left(\frac{2}{e^x+e^{-x}}\right)^2
$$

```python
def tanh_derivative(x):
  # Compute tanh first using your previously defined function
    t = tanh(x)

    # Apply derivative formula: 1 - tanh(x)^2
    return 1 - t**2
```

### Task: initialize the 1-hidden-layer parameters

Create a function that initializes the values of the weights `W1`, `W2` and biases `b1`, `b2`.

Use the provided `seed` so that results stay reproducible.

**NOTE**: make sure to re-run `GenerateAllWeights` each time you train the network (otherwise `train` function will have edited the matices and you want them to start random)


```python

def GenerateAllWeights(input_nodes, hidden_nodes, output_nodes, seed):
    np.random.seed(seed)

    W1 = np.random.randn(input_nodes, hidden_nodes)
    W2 = np.random.randn(hidden_nodes, output_nodes)

    b1 = np.zeros((1, hidden_nodes))
    b2 = np.zeros((1, output_nodes))

    return W1, b1, W2, b2
```

### Task: implement forward propagation for the 1-hidden-layer model

This function should:
- take one input example
- compute the hidden-layer values
- compute the output-layer values
- return the intermediate quantities needed later for gradient computation

 For example, the output of the 1st pass (input to hidden) could be called `z1` for before activation and `a1` = $(a_{11}, a_{12}, a_{13})$ for after activation. Likewise, the 2nd pass (output from the hidden to output) could be called `z2` before activation and `a2` = $(a_{21})$ after.


```python

def forward_propagation(x, activation_type):
  # Compute hidden layer pre-activation
    z1 = np.dot(x, W1) + b1

    # Apply chosen activation function to hidden layer
    a1 = ActivationFxn(z1, activation_type)

    # Compute output layer pre-activation
    z2 = np.dot(a1, W2) + b2

    # Use sigmoid at the output for binary classification
    a2 = sigmoid(z2)

    return z1, a1, z2, a2
```

### Task: Define an `output()` helper for batch prediction

Write a function that generates predictions for every row in a data matrix, `X`.

Your function should:

1. Loop over each row of `X`
2. Reshape that row into a column vector so it can be passed through the network
3. Call your `forward_propagation()` function on that sample
4. Catch all returned values from `forward_propagation()`
5. Save the final output from the output layer, `a2`, into a list

**Sanity check:** the number of outputs should match the number of rows in `X`.

```python
def output(X, activation_type):
  # Store model outputs for each row in X
    outputs = []

    # Loop through each example
    for i in range(X.shape[0]):
        # Reshape one row into a 1 x n_features matrix
        x = X[i].reshape(1, -1)

        # Run forward propagation and capture all returned values
        z1, a1, z2, a2 = forward_propagation(x, activation_type)

        # Save the final output prediction
        outputs.append(a2[0, 0])

    # Convert list to an n x 1 numpy array
    return np.array(outputs).reshape(-1, 1)
```

### Task: Compute gradients for the 1-hidden-layer model

Write a function that computes the gradients for all weights and biases in the 1-hidden-layer neural network.

The inputs to this function are the current sample, the true label, and the intermediate values from forward propagation. Use these values to compute the gradient terms needed to update the model parameters.

Your function should return:
- `dW1`
- `db1`
- `dW2`
- `db2`

For the gradient formulas, refer back to Labs 11 and 12.

```python
def gradient(x, y, a1, a2, z1, z2, activation_type):
  # Output layer error term
    dz2 = 2 * (a2 - y) * sigmoid_derivative(z2)

    # Gradients for W2 and b2
    dW2 = np.dot(a1.T, dz2)
    db2 = dz2

    # Hidden layer error term
    dz1 = np.dot(dz2, W2.T) * ActivationFxnDerivative(z1, activation_type)

    # Gradients for W1 and b1
    dW1 = np.dot(x.T, dz1)
    db1 = dz1

    return dW1, db1, dW2, db2
```

### Task: implement the 1-hidden-layer training loop

This is the function that brings the model together. The function should apply each of the helper functions you defined before in the correct order. The function should return the cost values for both the training and testing data in lists.

Across many epochs, it should:
- run forward propagation
- accumulate training cost
- accumulate gradients
- update parameters
- record train and test cost curves

Two details matter here:
- average the accumulated gradients by `x_train.shape[0]`
- report the **average cost per example** so your train and test curves are both shown as MSE and are comparable even though the train and test sets have different sizes


```python

def train(x_train, y_train, x_test, y_test, learning_rate, epochs, activation_type):
  global W1, b1, W2, b2

  # Lists to store average train and test cost each epoch
  train_costs = []
  test_costs = []

  # Number of training examples
  n_train = x_train.shape[0]

  for epoch in range(epochs):
      # Initialize gradient accumulators
      dW1_sum = np.zeros_like(W1)
      db1_sum = np.zeros_like(b1)
      dW2_sum = np.zeros_like(W2)
      db2_sum = np.zeros_like(b2)

      # Track total training cost for this epoch
      train_cost_total = 0

      # Loop through each training example
      for i in range(n_train):
          # Get one sample and one label
          x = x_train[i].reshape(1, -1)
          y = y_train[i].reshape(1, -1)

          # Forward propagation
          z1, a1, z2, a2 = forward_propagation(x, activation_type)

          # Add this sample's cost
          train_cost_total += cost(y, a2)

          # Compute gradients for this sample
          dW1, db1, dW2, db2 = gradient(x, y, a1, a2, z1, z2, activation_type)

          # Accumulate gradients
          dW1_sum += dW1
          db1_sum += db1
          dW2_sum += dW2
          db2_sum += db2

      # Average gradients over all training examples
      dW1_sum /= n_train
      db1_sum /= n_train
      dW2_sum /= n_train
      db2_sum /= n_train

      # Update parameters
      W1 -= learning_rate * dW1_sum
      b1 -= learning_rate * db1_sum
      W2 -= learning_rate * dW2_sum
      b2 -= learning_rate * db2_sum

      # Average training cost for this epoch
      avg_train_cost = train_cost_total / n_train
      train_costs.append(avg_train_cost)

      # Compute average test cost for this epoch
      test_preds = output(x_test, activation_type)
      avg_test_cost = cost(y_test, test_preds)
      test_costs.append(avg_test_cost)

  return train_costs, test_costs
```

##### Task: Create a function called `test()` for your 1-hidden-layer network.

For pedagogical consistency, we will use the name `test()` here.  
Unlike a print-only checkpoint, this version should **return a numeric accuracy value** so it can be reused in analysis, tables, and grading later.


```python
def test(inputs, outputs, activation_type):
   # Get predicted probabilities
    preds = output(inputs, activation_type)

    # Convert probabilities to class predictions using 0.5 threshold
    pred_classes = (preds >= 0.5).astype(int)

    # Compute accuracy
    accuracy = np.mean(pred_classes == outputs)

    return accuracy
    # return the accuracy as a float
```

### Task: Initialize parameters for the 2-hidden-layer model

Create the weight matrices and bias vectors needed for a network with two hidden transformations that returns three weights and biases.


```python
def GenerateAllWeights_2(input_nodes, first_hidden_nodes, second_hidden_nodes, output_nodes, seed):
    # Set seed for reproducibility
    np.random.seed(seed)

    # Initialize weights
    W1 = np.random.randn(input_nodes, first_hidden_nodes)
    W2 = np.random.randn(first_hidden_nodes, second_hidden_nodes)
    W3 = np.random.randn(second_hidden_nodes, output_nodes)

    # Initialize biases as zeros
    b1 = np.zeros((1, first_hidden_nodes))
    b2 = np.zeros((1, second_hidden_nodes))
    b3 = np.zeros((1, output_nodes))

    return W1, W2, W3, b1, b2, b3
```

- output layer

Make sure you return the intermediate values needed for backpropagation.


```python
def forward_propagation_2(x, activation_type):
  # First hidden layer: linear step
    z1 = np.dot(x, W1) + b1

    # First hidden layer: activation
    a1 = ActivationFxn(z1, activation_type)

    # Second hidden layer: linear step
    z2 = np.dot(a1, W2) + b2

    # Second hidden layer: activation
    a2 = ActivationFxn(z2, activation_type)

    # Output layer: linear step
    z3 = np.dot(a2, W3) + b3

    # Output layer uses sigmoid for binary classification
    a3 = sigmoid(z3)

    return z1, a1, z2, a2, z3, a3
```

### Task: Define the batch output helper for the 2-hidden-layer model

This helper should generate model outputs for a whole dataset, not just one example.


```python
def output_2(X, activation_type):
    outputs = []

    # Loop through each example
    for i in range(X.shape[0]):
        # Reshape row to (1, n_features)
        x = X[i].reshape(1, -1)

        # Forward pass through 2-layer network
        z1, a1, z2, a2, z3, a3 = forward_propagation_2(x, activation_type)

        # Store final output
        outputs.append(a3[0, 0])

    # Return as (n, 1) array
    return np.array(outputs).reshape(-1, 1)
```

### Task: Compute gradients for the 2-hidden-layer model

This is the deeper backpropagation step.

You now need to carry gradient information backward through:
- the output layer
- hidden layer 2
- hidden layer 1


```python
def gradient_2(x, y, a1, a2, a3, z1, z2, z3, activation_type):
    # Output layer error term
    dz3 = 2 * (a3 - y) * sigmoid_derivative(z3)

    # Gradients for W3 and b3
    dW3 = np.dot(a2.T, dz3)
    db3 = dz3

    # Backprop into second hidden layer
    dz2 = np.dot(dz3, W3.T) * ActivationFxnDerivative(z2, activation_type)

    # Gradients for W2 and b2
    dW2 = np.dot(a1.T, dz2)
    db2 = dz2

    # Backprop into first hidden layer
    dz1 = np.dot(dz2, W2.T) * ActivationFxnDerivative(z1, activation_type)

    # Gradients for W1 and b1
    dW1 = np.dot(x.T, dz1)
    db1 = dz1

    return dW1, db1, dW2, db2, dW3, db3
```

### Task: Implement the 2-hidden-layer training loop

This function is the main training engine for the rest of the project.

It should return the train and test cost curves so you can later compare stability, convergence, and possible overfitting.

As in Part 1:
- average the accumulated gradients by `x_train.shape[0]`
- report the **average cost per example** so your train and test curves are both shown as MSE


```python
def train_2(x_train, y_train, x_test, y_test, learning_rate, epochs, activation_type):
    # Correction carried into the notebook:
    # average by x_train.shape[0], not by X.shape[0]
    # TODO: write your code here
    global W1, b1, W2, b2, W3, b3

    # Store train and test cost for each epoch
    train_costs = []
    test_costs = []

    # Number of training examples
    n_train = x_train.shape[0]

    for epoch in range(epochs):
        # Initialize gradient accumulators
        dW1_sum = np.zeros_like(W1)
        db1_sum = np.zeros_like(b1)
        dW2_sum = np.zeros_like(W2)
        db2_sum = np.zeros_like(b2)
        dW3_sum = np.zeros_like(W3)
        db3_sum = np.zeros_like(b3)

        # Track total training cost for this epoch
        train_cost_total = 0

        # Loop through each training example
        for i in range(n_train):
            # Select one example and reshape to a row vector
            x = x_train[i].reshape(1, -1)
            y = y_train[i].reshape(1, -1)

            # Forward propagation
            z1, a1, z2, a2, z3, a3 = forward_propagation_2(x, activation_type)

            # Add this example's cost
            train_cost_total += cost(y, a3)

            # Compute gradients
            dW1, db1, dW2, db2, dW3, db3 = gradient_2(
                x, y, a1, a2, a3, z1, z2, z3, activation_type
            )

            # Accumulate gradients
            dW1_sum += dW1
            db1_sum += db1
            dW2_sum += dW2
            db2_sum += db2
            dW3_sum += dW3
            db3_sum += db3

        # Average gradients by number of training examples
        dW1_sum /= n_train
        db1_sum /= n_train
        dW2_sum /= n_train
        db2_sum /= n_train
        dW3_sum /= n_train
        db3_sum /= n_train

        # Update parameters
        W1 -= learning_rate * dW1_sum
        b1 -= learning_rate * db1_sum
        W2 -= learning_rate * dW2_sum
        b2 -= learning_rate * db2_sum
        W3 -= learning_rate * dW3_sum
        b3 -= learning_rate * db3_sum

        # Store average training cost
        train_costs.append(train_cost_total / n_train)

        # Compute and store test cost
        test_preds = output_2(x_test, activation_type)
        test_costs.append(cost(y_test, test_preds))

    return train_costs, test_costs
```

### Task: create a `test_2()` function for the 2-hidden-layer model

As before, this function should **return** a numeric accuracy value so it can be reused in analysis, comparison tables, and grading.


```python
def test_2(inputs, outputs, activation_type):
    # return the accuracy as a float

    # Get predicted probabilities from 2-layer network
    preds = output_2(inputs, activation_type)

    # Convert probabilities to class labels using 0.5 threshold
    pred_classes = (preds >= 0.5).astype(int)

    # Compute accuracy
    accuracy = np.mean(pred_classes == outputs)

    return accuracy
```

```python
def run_two_layer_experiment(first_hidden_nodes, second_hidden_nodes, activation_type, learning_rate, epochs, seed=42):
    global W1, W2, W3, b1, b2, b3

    # Initialize weights and biases for the 2-layer model
    W1, W2, W3, b1, b2, b3 = GenerateAllWeights_2(
        input_nodes, first_hidden_nodes, second_hidden_nodes, output_nodes, seed
    )

    start_time = time.time()

    # Train the model and get the train and test costs
    train_curve, test_curve = train_2(
        X_train, Y_train, X_test, Y_test,
        learning_rate, epochs, activation_type
    )

    elapsed = time.time() - start_time

    # Get the train and test accuracies
    train_accuracy = test_2(X_train, Y_train, activation_type)
    test_accuracy = test_2(X_test, Y_test, activation_type)

    # Return the results as a dictionary
    return {
        'activation': activation_type,
        'learning_rate': learning_rate,
        'epochs': epochs,
        'hidden_nodes_1': first_hidden_nodes,
        'hidden_nodes_2': second_hidden_nodes,
        'train_cost': train_curve[-1],
        'test_cost': test_curve[-1],
        'train_accuracy': train_accuracy,
        'test_accuracy': test_accuracy,
        'runtime': elapsed,
        'train_curve': train_curve,
        'test_curve': test_curve
    }
```

## Part 6: Comprehension questions

### Q1: What is the main difference between linear regression models and neural networks? Which pieces of the neural network make it more flexible?

**Answer:** The main difference is that linear regression models can only learn linear relationships, while neural networks can learn nonlinear and more complex patterns in the data. Linear regression makes a single weighted sum of the inputs, so it is limited to straight-line relationships between features and the output.

Neural networks are more flexible because they have multiple layers and activation functions. The hidden layers allow the model to transform the input data multiple times, and the activation functions (like tanh or ReLU) introduce nonlinearity. This means the network is not restricted to linear relationships and can capture more complicated patterns. Additionally, having multiple neurons in each layer increases the model’s capacity to learn different features at the same time, which makes it more powerful than a simple linear model.

### Q2: Why is tuning not the same thing as random guessing?

**Answer:**Tuning is not the same as random guessing because it is guided by evidence from previous runs, not trial-and-error with no direction. In this project, each change I made (activation, learning rate, hidden width) was based on what I observed in the results before, like cost curves and test accuracy.

For example, I first saw that tanh performed better than sigmoid and ReLU, so I kept tanh fixed and then tested different learning rates. After finding that 0.5 worked best, I moved on to testing hidden layer sizes. Each step had a clear reason and built on the previous results.

Random guessing would mean trying combinations without any reasoning or pattern, but tuning is a structured and iterative process where each experiment is designed to learn something and improve the model.

### Q3: Why can train accuracy and test accuracy tell different stories about model quality?

**Answer:** Train accuracy and test accuracy can tell different stories because they measure performance on different types of data. Train accuracy shows how well the model fits the data it was trained on, while test accuracy shows how well it generalizes to new, unseen data.

A model can have very high train accuracy but lower test accuracy if it has overfit the training data. This means it has learned patterns that are too specific to the training set and do not carry over to new data. On the other hand, if both train and test accuracy are low, the model is likely underfitting and not capturing the underlying patterns at all.

That is why test accuracy is usually more important for judging model quality, because it reflects how the model will perform in real-world situations, not just on the data it has already seen.
